```python

import re
import pandas as pd

from sklearn.feature_extraction.text import (
    TfidfVectorizer,
    ENGLISH_STOP_WORDS
)
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# 1. Демонстрационный датасет
data = [
    ("This movie was fantastic and amazing", "positive"),
    ("I really enjoyed this wonderful film", "positive"),
    ("The acting was excellent and the story was great", "positive"),
    ("Absolutely loved this movie", "positive"),
    ("A brilliant film with great performances", "positive"),
    ("The movie was fun and entertaining", "positive"),
    ("An excellent story and beautiful visuals", "positive"),
    ("I highly recommend this fantastic film", "positive"),
    ("The actors did a wonderful job", "positive"),
    ("A great movie with an interesting plot", "positive"),
    ("This film was enjoyable and impressive", "positive"),
    ("One of the best movies I have watched", "positive"),
    ("The story was touching and beautifully written", "positive"),
    ("Amazing performances made this film special", "positive"),
    ("I liked every minute of this movie", "positive"),
    ("This movie was terrible and boring", "negative"),
    ("I hated this awful film", "negative"),
    ("The acting was poor and the story was bad", "negative"),
    ("Absolutely disliked this movie", "negative"),
    ("A horrible film with terrible performances", "negative"),
    ("The movie was dull and disappointing", "negative"),
    ("A boring story and ugly visuals", "negative"),
    ("I cannot recommend this awful film", "negative"),
    ("The actors did a terrible job", "negative"),
    ("A bad movie with a confusing plot", "negative"),
    ("This film was annoying and disappointing", "negative"),
    ("One of the worst movies I have watched", "negative"),
    ("The story was dull and poorly written", "negative"),
    ("Terrible performances made this film awful", "negative"),
    ("I disliked every minute of this movie", "negative"),
]

df = pd.DataFrame(data, columns=["text", "label"])

# 2. Очистка текста
# Сохраняем not, no и nor, так как отрицания важны
# для определения тональности отзыва.
stop_words = set(ENGLISH_STOP_WORDS) - {"not", "no", "nor"}

def clean_text(text):
    # Приведение к нижнему регистру
    text = text.lower()

    # Удаление цифр и пунктуации
    text = re.sub(r"[\d]+", " ", text)
    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"_+", " ", text)

    # Токенизация: разделение на слова
    tokens = text.split()

    # Удаление стоп-слов
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    return " ".join(tokens)

# Проверка очистки
df["clean_text"] = df["text"].apply(clean_text)

print("Пример очистки:")
print("До:", df["text"].iloc[0])
print("После:", df["clean_text"].iloc[0])

# 3. Разделение данных на обучающую и тестовую выборки
X_train, X_test, y_train, y_test = train_test_split(
    df["text"],
    df["label"],
    test_size=0.3,
    random_state=42,
    stratify=df["label"]
)

# 4. TF-IDF векторизация
vectorizer = TfidfVectorizer(
    preprocessor=clean_text,
    tokenizer=str.split,
    token_pattern=None,
    lowercase=False,
    min_df=1,
    max_df=0.9,
    sublinear_tf=True
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("\nРазмер обучающей матрицы:", X_train_tfidf.shape)
print("Размер словаря:", len(vectorizer.vocabulary_))

# 5. Обучение классификатора
model = LogisticRegression(max_iter=1000)
model.fit(X_train_tfidf, y_train)

# 6. Предсказания и оценка качества
y_pred = model.predict(X_test_tfidf)

print("\nAccuracy:", round(accuracy_score(y_test, y_pred), 3))
print("\nClassification report:")
print(classification_report(y_test, y_pred, zero_division=0))

print("Confusion matrix:")
print(confusion_matrix(
    y_test,
    y_pred,
    labels=["negative", "positive"]
))

# 7. Анализ ошибок
errors = pd.DataFrame({
    "text": X_test.values,
    "actual": y_test.values,
    "predicted": y_pred
})

errors = errors[errors["actual"] != errors["predicted"]]

print("\nКоличество ошибок:", len(errors))
print("\nПервые пять ошибок:")
print(errors.head(5).to_string(index=False))

if len(errors) < 5:
    print(
        "\nВ тестовой выборке меньше пяти ошибок. "
        "Для анализа пяти ошибок используйте более крупный "
        "реальный датасет или другую тестовую выборку."
    )
```
